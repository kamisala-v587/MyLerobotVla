### 2. norm stats计算

```python
import json
from pathlib import Path

def norm_stats(data_path):
    data_path  = Path(data_path)
    STATS_JSON = data_path / "meta/stats.json"
    OUTPUT_JSON = data_path / "norm_stats.json"  # 数据集根目录；若要写在 meta/ 下则改为 META_DIR / "norm_stats.json"
    MAPPING = {
        "observation.state": "state",
        "action": "actions",
    }
    KEEP = ("mean", "std", "q01", "q99")
    with STATS_JSON.open(encoding="utf-8") as f:
        src = json.load(f)
    out = {
        "norm_stats": {
            dst: {k: src[src_key][k] for k in KEEP}
            for src_key, dst in MAPPING.items()
        }
    }
    with OUTPUT_JSON.open("w", encoding="utf-8") as f:
        json.dump(out, f, indent=2)
        f.write("\n")
    print("done:", OUTPUT_JSON)

norm_stats('/vla/.data/test')
```
