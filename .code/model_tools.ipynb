```python
import torch
from safetensors.torch import save_file

src = "/vla/my_vla/policy/checkpoints/global_step_8000/actor/model_state_dict/full_weights.pt"
dst = "/vla/my_vla/policy/checkpoints/global_step_8000/actor/model_state_dict/model.safetensors"

state = torch.load(src, map_location="cpu", weights_only=True)

# 可选：去掉与 lm_head 共享的 embed_tokens（base 模型 safetensors 里没有这 1 个 key）
extra = "paligemma_with_expert.paligemma.model.language_model.embed_tokens.weight"
state.pop(extra, None)

save_file(state, dst)
print(f"saved {len(state)} tensors -> {dst}")
```
