```python
from lerobot.datasets.lerobot_dataset import LeRobotDataset
ds = LeRobotDataset('/vla/.data/adjust_bottle')
ds
```

```python
import numpy as np
import matplotlib.pyplot as plt

def show_sample_images(ds, idx, w=3):
    s, keys = ds[idx], ds.meta.camera_keys
    fig, axes = plt.subplots(1, len(keys), figsize=(w * len(keys), w))
    for ax, k in zip(np.atleast_1d(axes), keys):
        ax.imshow(np.asarray(s[k]).transpose(1, 2, 0))
        ax.set_title(k.rsplit(".", 1)[-1], fontsize=9)
        ax.axis("off")
    fig.tight_layout()

show_sample_images(ds, 50)
```
